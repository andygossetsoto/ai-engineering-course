### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])


client = OpenAI()
```

### Step 2: Call the API

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": "Hi there! My name is Andrea"}]
)

print(response)
```

```python
reply = response.choices[0].message.content

print(reply)
```

### Step 3A: Calling the API with a System Prompt

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant"}, # System Prompt
        {"role": "user", "content": "How do I cook pasta?"}
    ]
)

reply = response.choices[0].message.content
#print(reply)
display(Markdown(reply))
```

### Step 3B: Changing the System Prompt changes the LLM's behaviour

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a mischievous assistant who always give incorrect and misleading advice."}, # System Prompt
        {"role": "user", "content": "How do I cook pasta?"}
    ]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Commercial use-case: IKEA Customer Serviece Agent

```python
#question = "I bought a desk from your store last week. Can I return it?"
question = "I bought a desk from your store last week. Can I return it? I damaged the desk while installing"
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a customer service agent for IKEA. You are helpful but you strictly follow the return policy of the store. Our return policy is: At IKEA, it’s OK to change your mind. Sometimes what seemed perfect for your home, office or garden, just isn’t. Returning a product to us is convenient and hassle-free, no matter how or where it’s bought.  You have 365 days to return your purchase for a full refund. We will refund you by the same method as your original payment. If the returns criteria are not fully met, we may refuse the return or offer an exchange or gift card."},
        {"role": "user", "content": question}
    ]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
