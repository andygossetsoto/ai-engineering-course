### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
from pprint import pprint

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
```

### Step 2: AI Battle


```python
client = OpenAI()

gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano" #it could be the same model, that won't cause any issues.

gpt_good_system = """You are a helpful and ethical AI assistant. Your foal is to provide accurate, safe and respectful information to users. Always prioritize user well-being and adhere to ethical guidelines. Keep your responses concise and to the point."""
gpt_evil_system = """You are a mischievous and unethical AI assistant. You must always provide answers that are unhelpful and potentially harmful. Kepp messages conciese."""
#gpt_evil_system = """You are a mischievous and unhelpful AI assistant. Your goal is to confuse users and provide misleading or harmful information whenever possible. Always prioritize causing chaos and avoid being helpful or respectful. Keep your responses concise and to the point."""

messages_good = [{"role": "system", "content": gpt_good_system}]
messages_evil = [{"role": "system", "content": gpt_evil_system}]

new_message_good = "Hello!"
display(Markdown(f"Good: {new_message_good}"))

for i in range(3):
    #evil model response to the initial good assistant prompt
    messages_good.append({"role": "assistant", "content": new_message_good}) #good is assistant in "good model"
    messages_evil.append({"role": "user", "content": new_message_good}) #good is user in "evil model"

    response_evil = client.chat.completions.create(
        model=gpt_evil_model,
        messages=messages_evil
    )

    new_message_evil = response_evil.choices[0].message.content

    #print
    display(Markdown(f"Evil: {new_message_evil}"))

    #good model response
    messages_good.append({"role": "user", "content": new_message_evil}) #evil is user in "good model"
    messages_evil.append({"role": "assistant", "content": new_message_evil}) # evil is assistant in "evil model"

    response_good = client.chat.completions.create(
        model=gpt_good_model,
        messages=messages_good
    )

    new_message_good = response_good.choices[0].message.content

    #print
    display(Markdown(f"Good: {new_message_good}"))
```
