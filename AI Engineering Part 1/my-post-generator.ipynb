### Step 1: Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
```

### Step 2: Searching the web

```python
client = OpenAI()
#gpt-3.5-turbo
#text-embedding-ada-002
completion = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        #{"role":"user", "content": "What is the date today? Tell me the name of 5 famous people that have their birthdays today and one fun fact about them"}
        {"role":"user", "content": "What is the date today? What are todays latest AI-driven approaches to improving web accessibility (a11y) and perfomance optimization for front-end? Give me exact names of companies or products involved in this. All the options should have AI and you should explain how AI plays a role in the company/product."}
        #{"role":"user", "content": "What is the date today? What new AI regulations or guidelines affecting web development and data privacy were announced this month?"}
    ],
    max_tokens=500
)
```

```python
search_results = completion.choices[0].message.content
display(Markdown(search_results))
```

### Step 3: Summarize and write social media post

```python
system_message = """
You are a helpful assistant who creates amazing linkedin post about AI based on 
text that you are provided with. Make the posts engaging and interesting.
Also make concise and add a thought provoking question at the end to encourage comments.
"""

prompt = f"""
The following text contains top accessibility solutions with AI for web pages.
Now please write my linkedin post based on this information.

{search_results}

"""

completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role":"system", "content": system_message},
        {"role":"user", "content": prompt}
    ],
    max_tokens=500
)

chat_response = completion.choices[0].message.content
```

```python
print(chat_response)
```
