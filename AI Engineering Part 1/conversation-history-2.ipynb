### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])


client = OpenAI()
```

### Step 2: Call the API with manual context

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Next week I'm travelling to Spain."},
    {"role": "assistant", "content": "That’s exciting! Do you need any help with planning your trip, like places to visit, transportation tips, or packing suggestions?"},
    {"role": "user", "content": "Where am I travelling to next week?"},
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Call the API with Appended context

```python
from pprint import pprint

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Next week I'm travelling to Spain."},
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = response.choices[0].message.content
messages.append({"role":"assistant", "content": reply})
pprint(messages)
```
